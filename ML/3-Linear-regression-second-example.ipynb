#### 3.1 Linear Regression

````python
from sklearn.linear_model  import LinearRegression

#model definition
l_reg=LinearRegression()
````

#### 3.2. Random Forest Regressor

````python
from sklearn.ensemble import RandomForestRegressor

# model definition
rfr = RandomForestRegressor(n_estimators=30,random_state=42)
````

### 4. Model Training

#### 4.1. Linear Regression

````python
# train the model
l_reg.fit(X_train,y_train)
````

````python
# The model parameters
l_reg.get_params()
````

````python
## Let's see the if the model got good learning
print(f"The accuracy after learning %2.3f Percent"%(l_reg.score(X_train,y_train)*100))
````

````python
## What about the model coef ?

print(f"The model coefficients are: \n")
lr_coefs=pd.DataFrame(l_reg.coef_, columns=['Coefficients'], index=X_train.columns)

new_line = pd.Series({'Coefficients':l_reg.intercept_}, name='Intercept')
lr_coefs = pd.concat([lr_coefs, new_line.to_frame().T])
lr_coefs.plot(kind='barh', figsize=(9, 7))
plt.title('Linear Regression model')
plt.axvline(x=0, color='.5')
plt.subplots_adjust(left=.3)
plt.show()
````

#### 4.2. Random forest

````python
# Train the RF model
rfr.fit(X_train,y_train)
````

````python
rfr.get_params()
````

````python
## Let's see the if the model got good learning
print(f"The accuracy after learning %2.3f Percent"%(rfr.score(X_train,y_train)*100))
````

````python
## What about the model feature importance ?

print(f"The model coefficients are: \n")
rfr_f_imp=pd.DataFrame(rfr.feature_importances_*100, columns=['Features Importances (%)'], index=X_train.columns)

# plotly package could also be use for nice display
import plotly.express as px

fig = px.bar(rfr_f_imp, x='Features Importances (%)', orientation='h',height=500, width=800)
fig.show()
````

### 6. Model parameter tuning

#### When you are not satisfy with the accuracy of your model, this step consist to go back to your model to change the model hyperparameters or the data splitting. 

##### `Scikit-learn` propose some modules to handle this type of problem 

   - `Hyper-parameter optimizers`  [Click here for more details](https://scikit-learn.org/stable/modules/grid_search.html?highlight=hyper+parameter+optimizers)
   - `Model validation` [Click here for more details](https://scikit-learn.org/stable/modules/cross_validation.html)
   


````python
# Manual Example for random forest model
estimator=0
score_=0
for i in range(2,300):
    # model defenition
    rfr = RandomForestRegressor(n_estimators=i,random_state=42)
    # Train the RF model
    rfr.fit(X_train,y_train)
    rfr_pred = rfr.predict(X_test)
    if metrics.r2_score(y_test, rfr_pred)>score_:
        score_=metrics.r2_score(y_test, rfr_pred)
        estimator=i
        
print(f"The best estimator %2d, Accuracy_Score = %2.2f" % (estimator, score_))
````
